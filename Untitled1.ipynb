```python
# Install libraries
!pip install transformers torch accelerate --quiet

# Imports
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Model name
model_name = "EleutherAI/gpt-neo-125M"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Set pad token
tokenizer.pad_token = tokenizer.eos_token

# GPU or CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

# Text generation function
def generate_text(prompt, max_length=100, temperature=0.8, top_k=50, top_p=0.9):

    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    output = model.generate(
        **inputs,
        max_length=max_length,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        repetition_penalty=1.2,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    return tokenizer.decode(output[0], skip_special_tokens=True)

# Experiment function
def experiment(prompt):

    print("\n--- Experimenting with Parameters ---\n")

    settings = [
        {"temperature": 0.3, "top_k": 20, "top_p": 0.7},
        {"temperature": 0.7, "top_k": 50, "top_p": 0.9},
        {"temperature": 1.0, "top_k": 100, "top_p": 0.95},
    ]

    for i, s in enumerate(settings):

        print(f"\nSetting {i+1}: {s}")

        text = generate_text(
            prompt,
            temperature=s["temperature"],
            top_k=s["top_k"],
            top_p=s["top_p"]
        )

        print(text)
        print("-" * 60)

# Main program
prompt_text = "Once upon a time there was a beautiful queen"

print("\nGenerated Text:\n")
print(generate_text(prompt_text))

experiment(prompt_text)
```

```python
# Install first:
# !pip install transformers sentencepiece sacremoses torch --quiet

from transformers import MarianMTModel, MarianTokenizer
import torch

# Supported language models
language_models = {
    "spanish": "Helsinki-NLP/opus-mt-en-es",
    "french": "Helsinki-NLP/opus-mt-en-fr",
    "hindi": "Helsinki-NLP/opus-mt-en-hi",
    "german": "Helsinki-NLP/opus-mt-en-de",
    "italian": "Helsinki-NLP/opus-mt-en-it"
}

# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# Store models and tokenizers
models = {}
tokenizers = {}

# Load models
for lang, model_name in language_models.items():

    print(f"Loading {lang} model...")

    tokenizers[lang] = MarianTokenizer.from_pretrained(model_name)

    models[lang] = MarianMTModel.from_pretrained(model_name).to(device)

print("All models loaded successfully!")

# Translation function
def translate_text(user_text, target_language):

    target_language = target_language.lower()

    if target_language not in models:
        return f"Translation to {target_language} not supported."

    tokenizer = tokenizers[target_language]
    model = models[target_language]

    inputs = tokenizer(
        [user_text],
        return_tensors="pt",
        padding=True
    ).to(device)

    translated = model.generate(**inputs)

    output = tokenizer.batch_decode(
        translated,
        skip_special_tokens=True
    )[0]

    return output

# Main program
print("\n--- Multilingual Translator ---")

print("Supported languages:")
print(", ".join(language_models.keys()))

while True:

    text = input("\nEnter text: ")

    if text.lower() == "exit":
        print("Exiting translator...")
        break

    lang = input("Translate to: ")

    result = translate_text(text, lang)

    print("Translated:", result)
```

```python


import torch
from diffusers import StableDiffusionPipeline
from IPython.display import display

model_id = "sd2-community/stable-diffusion-2-1"
device = "cuda" if torch.cuda.is_available() else "cpu"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
).to(device)

prompt = input("Enter your image prompt: ")
neg = input("Enter negative prompt (optional): ")

styles = [
    "oil painting style",
    "anime style",
    "cinematic lighting"
]

print("\n--- Generating Images with Different Styles ---\n")

for i, style in enumerate(styles):

    styled_prompt = f"{prompt}, {style}"

    image = pipe(
        styled_prompt,
        negative_prompt=neg if neg else None,
        num_inference_steps=30,
        guidance_scale=7.5
    ).images[0]

    filename = f"output_{i+1}.png"

    image.save(filename)

    print(f"Style: {style}")

    display(image)
```

```python

# 1️⃣ Install required libraries
!pip install pretty_midi numpy midi2audio --quiet

# 2️⃣ Imports
import pretty_midi
import numpy as np
from IPython.display import Audio, display
import os

# 3️⃣ Function to generate and play MIDI
def create_and_play_midi():
    # Create PrettyMIDI object
    midi = pretty_midi.PrettyMIDI()

    # Define instruments (Piano, Violin, Flute)
    instruments = [
        pretty_midi.Instrument(program=0),   # Piano
        pretty_midi.Instrument(program=40),  # Violin
        pretty_midi.Instrument(program=73)   # Flute
    ]

    # Generate random notes for each instrument
    for idx, instrument in enumerate(instruments):
        for i in range(10):
            pitch = np.random.randint(60, 72)
            start = i * 0.5
            end = start + 0.5
            note = pretty_midi.Note(
                velocity=90,
                pitch=pitch + idx * 3,   # slight variation per instrument
                start=start,
                end=end
            )
            instrument.notes.append(note)
        midi.instruments.append(instrument)

    # Save MIDI file
    midi_file = "generated_music.mid"
    midi.write(midi_file)
    print(f"🎶 Multi-instrument music generated: '{midi_file}' 🎶")


# 4️⃣ Run the function
create_and_play_midi()

```

```python
import matplotlib.pyplot as plt

# Load the generated MIDI file
midi_data = pretty_midi.PrettyMIDI('generated_music.mid')

# Prepare data for plotting
notes_to_plot = []
for instrument in midi_data.instruments:
    for note in instrument.notes:
        notes_to_plot.append({
            'start': note.start,
            'end': note.end,
            'pitch': note.pitch,
            'instrument': instrument.program
        })

# Create the piano roll plot
plt.figure(figsize=(15, 6))

# Assign a unique color for each instrument
colors = plt.cm.get_cmap('tab10', len(midi_data.instruments))

for i, instrument in enumerate(midi_data.instruments):
    inst_notes = [n for n in notes_to_plot if n['instrument'] == instrument.program]
    for note in inst_notes:
        plt.barh(note['pitch'], note['end'] - note['start'], left=note['start'], height=0.8, color=colors(i), label=f'Instrument {instrument.program}' if note == inst_notes[0] else "")

plt.xlabel('Time (s)')
plt.ylabel('MIDI Note Number')
plt.title('MIDI Piano Roll Visualization')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.yticks(np.arange(min(n['pitch'] for n in notes_to_plot) - 2, max(n['pitch'] for n in notes_to_plot) + 3, 1.0))

# Add legend if there are multiple instruments with unique labels
handles, labels = plt.gca().get_legend_handles_labels()
unique_labels = dict(zip(labels, handles))
plt.legend(unique_labels.values(), unique_labels.keys())

plt.tight_layout()
plt.show()
```

```python
!pip install transformers accelerate torch --quiet

from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "bigcode/starcoder2-3b"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

# Ask user only at runtime
user_task = input("Enter the programming task (example: Python program to check prime number): ")

system_instruction = """
You are a helpful programming assistant.
Generate ONLY the required program.
Do NOT add explanations, comments, extra text, or unrelated code.
Write clean, correct, and minimal code.
"""

prompt = f"{system_instruction}\nTask: {user_task}\nCODE:\n"

inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

output = model.generate(
    **inputs,
    max_new_tokens=160,
    temperature=0.1,
    do_sample=False,                     # deterministic
    top_p=0.9,
    repetition_penalty=1.05,
    eos_token_id=tokenizer.eos_token_id,
    pad_token_id=tokenizer.eos_token_id
)

result = tokenizer.decode(output[0], skip_special_tokens=True)

# Extract only generated code after "CODE:"
if "CODE:" in result:
    result = result.split("CODE:")[1]

print("\nGenerated Code:\n")
print(result.strip())
```

```python
# ---------------- INSTALL DEPENDENCIES ----------------
!pip install transformers sentence-transformers faiss-cpu pypdf python-docx --quiet

# ---------------- IMPORTS ----------------
import numpy as np
import faiss
import torch
import re

from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from google.colab import files

# ---------------- UPLOAD FILE ----------------
print("Upload your PDF or DOCX file:\n")

uploaded = files.upload()

# Get uploaded file name
file_name = list(uploaded.keys())[0]

print(f"\nUploaded File: {file_name}")

# ---------------- FILE READING FUNCTION ----------------
def read_file(fname):

    # DOCX
    if fname.endswith(".docx"):

        from docx import Document

        document = Document(fname)

        text = ""

        for para in document.paragraphs:
            text += para.text + " "

        return text

    # PDF
    elif fname.endswith(".pdf"):

        from pypdf import PdfReader

        reader = PdfReader(fname)

        text = ""

        for page in reader.pages:

            page_text = page.extract_text()

            if page_text:
                text += page_text + " "

        return text

    # Unsupported
    else:
        return ""

# ---------------- READ DOCUMENT ----------------
text = read_file(file_name)

# ---------------- VALIDATION ----------------
if not text or not text.strip():

    print("File is empty or unsupported.")

    raise SystemExit

# ---------------- CHUNKING ----------------
chunk_size = 400

chunks = []

for i in range(0, len(text), chunk_size):

    chunk = text[i:i + chunk_size]

    chunks.append(chunk)

print(f"\nTotal Chunks Created: {len(chunks)}")

# ---------------- LOAD EMBEDDING MODEL ----------------
print("\nLoading embedding model...")

embedder = SentenceTransformer("all-mpnet-base-v2")

# ---------------- CREATE EMBEDDINGS ----------------
print("Creating embeddings...")

embeddings = embedder.encode(chunks)

embeddings = np.array(embeddings).astype("float32")

# ---------------- CREATE FAISS INDEX ----------------
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("FAISS index created successfully!")

# ---------------- LOAD FLAN-T5 MODEL ----------------
print("\nLoading FLAN-T5 model...")

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")

model = AutoModelForSeq2SeqLM.from_pretrained(
    "google/flan-t5-base"
)

# ---------------- DEVICE ----------------
device = "cuda" if torch.cuda.is_available() else "cpu"

model = model.to(device)

print(f"Using device: {device}")

# ---------------- GENERATION FUNCTION ----------------
def generate_answer(prompt):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=120,
        do_sample=False
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer

# ---------------- CHAT LOOP ----------------
print("\nRAG Chatbot Ready!")
print("Type 'exit' to stop.\n")

while True:

    question = input("Ask your question: ")

    # Exit
    if question.lower() == "exit":

        print("Chatbot exited.")

        break

    # ---------------- QUESTION EMBEDDING ----------------
    q_embed = embedder.encode([question])

    q_embed = np.array(q_embed).astype("float32")

    # ---------------- SEARCH ----------------
    top_k = 5

    distances, indices = index.search(q_embed, top_k)

    # ---------------- RETRIEVE CONTEXT ----------------
    retrieved_chunks = []

    for idx in indices[0]:

        retrieved_chunks.append(chunks[idx])

    context_raw = " ".join(retrieved_chunks)

    # ---------------- CLEAN CONTEXT ----------------
    context = re.sub(
        r"http\S+|DOI:\s*\d+|\[\d+\]",
        "",
        context_raw
    )

    context = " ".join(context.split())

    # ---------------- PROMPT ----------------
    prompt = f"""
Answer the question using ONLY the given context.

Context:
{context}

Question:
{question}

Answer:
"""

    # ---------------- GENERATE ANSWER ----------------
    answer = generate_answer(prompt)

    answer = answer.strip()

    # ---------------- FALLBACK ----------------
    if len(answer) < 10:

        answer = "Answer not found in the provided document."

    # ---------------- OUTPUT ----------------
    print("\nAnswer:\n")

    print(answer)

    print("\n" + "-" * 60 + "\n")
```
